### Online Shoppers Purchasing Intention Dataset

The Online Shoppers Purchasing Intention Dataset captures user behavior on an e-commerce website to predict whether a session results in a purchase. Key numeric metrics include the number of pages visited in different categories: "Administrative", "Informational", and "Product Related", which count how many pages a user browsed in each category, with higher numbers generally indicating greater engagement. Additionally, there are corresponding duration metrics such as "Administrative Duration", "Informational Duration", and "Product Related Duration" that measure the total time spent on those pages in seconds. The longer durations may suggest either interest or difficulty navigating. "Bounce Rate" and "Exit Rate" are fractional values representing the likelihood of leaving pages quickly or ending the session on a page, with higher values typically indicating less engagement. "PageValues" estimates the monetary contribution of pages to revenue, where higher values suggest pages visited by users more likely to purchase. "SpecialDay" measures how close the session occurred to a holiday or special event on a scale from 0 to 1, with values near 1 signifying major shopping days. 

Categorical features include "Month" (month of the session), "OperatingSystems", "Browser", "Region", and "TrafficType" (all int-coded), as well as "VistorType" (new or returning) and "Weekend" (whether the session occurred on a weekend). The target variable, "Revenue", is a boolean indicating whether the session ended in a purchase.

This dataset only includes visitors to a specific e-commerce site, so findings may not generalize to all online shoppers. Sessions likely overrepresent engaged users, while casual visitors may leave without any meaningful interactions. Duration metrics may overestimate attention if users leave the site open when distracted. Additionally, since only a portion of sessions result in purchases, the dataset may be imbalanced, which is important to account for in predictive modeling.

### E-Commerce Customer Behaviour Dataset 

The E-Commerce Customer Behaviour Dataset is a synthetic dataset generated for educational purposes by Gretal AI. It simulates how customers interact with an online retail platform. Each record represents a unique customer identified by a "Customer ID", an integer label used for identification. "Age" is expressed in years and reflects the customer's age, useful to capture how purchasing behaviour might differ across age groups. "Gender" is a categorical text label (e.g., Male, Female, Other) used to categorize customers for group comparisons. "Location" is a categorical descriptor giving the customer's geographic area, which can help explore regional behaviour differences. "Annual Income" is in US dollars and shows the customer's income level. It helps us segment customers by spending capacity. The dataset includes "Purchase History", which is a multi-valued list of past transactions containing dates, product categories, and prices in USD, giving us a depiction of what and when customers purchased. "Browsing History" logs the products viewed and associated timestamps, capturing site engagement even when a purchase doesn't occur. "Product Reviews" consists of free-text review content and a rating on a 1-5 star scale, representing qualitative user sentiment toward purchased items. "Time on Site" measures the total time in minutes that a customer spent during their visit; larger values indicate deeper engagement or more extensive browsing before deciding on a purchase.

Because this dataset was generated synthetically using Gretal AI to imitate realistic e-commerce behaviour, it may not capture all irregularities found in real user interaction data, which can limit how well models built on it generalize to real systems. The features are designed to look realistic, but these are simulated values rather than observed ones, so patterns reflect generation assumptions more than true behaviours. Additionally, since the dataset includes only a few dozen customers and is small, findings may be unstable.

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd
import numpy as np
import ast
df = pd.read_csv('data/00-raw/E_Commerce_Customer_Behavior_Dataset.csv')
#The data is already tidy, every measurable variable has a separate column, every observation is in a different row
#There is a unique identifier but it is wrong, the data is rectangular
df['Customer ID'] = np.arange(len(df))
size = df.size
length = len(df)
#Size of the dataset is 450, with 50 rows
df.isnull().any()
#There is no missing data at all
df.dtypes
#All data types are expected
df['Purchase History'] = df['Purchase History'].apply(ast.literal_eval)

df = df.explode('Purchase History')

df = df.join(pd.json_normalize(df['Purchase History']))
df.drop(columns=['Product Category', 'Purchase Date'], inplace=True)
#Makes each purchase a separate column
df.describe()
#All min/max seem reasonable, no outliers seem to be present
#Product Reviews seem to only apply for the first product, cannot be considered for other products
categories = []

for purchase in df['Purchase History']:
    if isinstance(purchase, dict):
        if 'Category' in purchase:
            categories.append(purchase['Category'])
        elif 'Product Category' in purchase:
            categories.append(purchase['Product Category'])
        else:
            categories.append(np.nan)
    else:
        categories.append(np.nan)
df['Category'] = categories   
#Categories were mislabeled
times = []
for purchase in df['Purchase History']:
    if isinstance(purchase, dict):
        if 'Date' in purchase:
            times.append(purchase['Date'])
        elif 'Purchase Date' in purchase:
            times.append(purchase['Purchase Date'])
        else:
            times.append(np.nan)
    else:
        times.append(np.nan)
df['Date'] = times
#Times were mislabeled
df.to_csv('data/02-processed/E_Commerce_Customer_Behavior_Dataset_Cleaned.csv', index=False)
```
